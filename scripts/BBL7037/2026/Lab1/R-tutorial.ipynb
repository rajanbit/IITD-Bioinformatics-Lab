```r
# Loading package
library("ggplot2")
```

```r
# Creating data
x = c(0, 1, 2, 3, 4)
y = c(2, 5, 7, 10, 9)
```

```r
# Print vector x
print(x)
```

```r
# Creating dataframe
df <- data.frame(val1 = x, val2 = y)

# Print df
print(df)
```

```r
# Plotting scatter plot
ggplot(df, aes(x=val1, y=val2)) + geom_point(size = 3, color = "blue")
```

```r
# Plotting line chart
ggplot(df, aes(x=x, y=y)) + geom_line(size = 3, color = "blue")
```
